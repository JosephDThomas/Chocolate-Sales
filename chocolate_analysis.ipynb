## Extracting the data for all three shops

```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np

store1 = pd.read_csv("store_1.csv", sep=",", header=0)
store2 = pd.read_csv("store_2.csv", sep=",", header=0)
store3 = pd.read_csv("store_3.csv", sep=",", header=0)
```

```python
from scipy import stats 

#Is choc chunk normally distributed for all stores?
print("For store 1 ",stats.shapiro(store1.iloc[:,0]))
print("For store 2 ", stats.shapiro(store2.iloc[:,0]))
print("For store 3 ", stats.shapiro(store3.iloc[:,0]))
```

```python
from scipy import stats 

#Is milky travel normally distributed for all stores?
print("For store 1 ",stats.shapiro(store1.iloc[:,1]))
print("For store 2 ", stats.shapiro(store2.iloc[:,1]))
print("For store 3 ", stats.shapiro(store3.iloc[:,1]))
```

```python
from scipy import stats 

#Is chunkie normally distributed for all stores?
print("For store 1 ",stats.shapiro(store1.iloc[:,2]))
print("For store 2 ", stats.shapiro(store2.iloc[:,2]))
print("For store 3 ", stats.shapiro(store3.iloc[:,2]))
```

### All chocolate bars for all stores appear to be normally distributed

```python
#Using ANOVA test for Choc chunk across store
stats.f_oneway(store1.iloc[:,0], store2.iloc[:,0], store3.iloc[:,0])
```

```python
#Using ANOVA test for Milky travel across store
stats.f_oneway(store1.iloc[:,1], store2.iloc[:,1], store3.iloc[:,1])
```

```python
#Using ANOVA test for Chunkie across store
stats.f_oneway(store1.iloc[:,2], store2.iloc[:,2], store3.iloc[:,2])
```

#### There are plans to introduce a new chocolate bar in some stores. Due to shelf restrictions one of the chocolate bars in each of the stores must be removed to account for this.

#### It is decided store 2 will remove choc chunk and store 3 will remove milky travel replacing them with the new chocolate bar honey bite. It is decided to keep chunkie in store 1 as it is near store 2 to capture sales that would have been at store 2 instead while allowing room for the new chocolate bar at store 2.



```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np

store1 = pd.read_csv("store_1_new.csv", sep=",", header=0)
store2 = pd.read_csv("store_2_new.csv", sep=",", header=0)
store3 = pd.read_csv("store_3_new.csv", sep=",", header=0)
```

```python
from scipy import stats 

print("Is choc chunk normally distributed for the stores?")

a, b = stats.shapiro(store1["choc chunk"])
c, d = stats.shapiro(store3["choc chunk"])
if a > 0.05 and c > 0.05:
    print("yes")
else:
    print("no")

print("Is milky travel normally distributed for the stores?")
a,b = stats.shapiro(store1["milky travel"])
c, d = stats.shapiro(store2["milky travel"])
if a > 0.05 and c > 0.05:
    print("yes")
else:
    print("no")

print("Is chunkie normally distributed for the stores?")
a,b = stats.shapiro(store1["chunkie"])
c,d = stats.shapiro(store2["chunkie"])
if a > 0.05 and c > 0.05:
    print("yes")
else:
    print("no")

print("Is honey bite normally distributed for the stores?")
a,b = stats.shapiro(store2["honey bite"])
c,d = stats.shapiro(store3["honey bite"])
if a > 0.05 and c > 0.05:
    print("yes")
else:
    print("no")
```

```python
from matplotlib import pyplot as plt 

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

#For Choc chunk
axes[0,0].hist(store1["choc chunk"], label="Store 1")
axes[0,0].hist(store3["choc chunk"], alpha=0.6, label="Store 3")
axes[0,0].legend()
axes[0,0].set_title(label="Choc chunk sales")

print("Mean of store 1 choc chunk sales ",np.mean(store1["choc chunk"]))
print("Mean of store 3 choc chunk sales ", np.mean(store3["choc chunk"]))

#For Milky travel
axes[0,1].hist(store1["choc chunk"], label="Store 1")
axes[0,1].hist(store3["choc chunk"], alpha=0.6, label="Store 2")
axes[0,1].legend()
axes[0,1].set_title(label="Milky travel sales")

print("Mean of store 1 milky travel sales ",np.mean(store1["milky travel"]))
print("Mean of store 2 milky travel sales ", np.mean(store2["milky travel"]))

#For Chunkie
axes[1,0].hist(store1["chunkie"], label="Store 1")
axes[1,0].hist(store3["chunkie"], alpha=0.6, label="Store 2")
axes[1,0].legend()
axes[1,0].set_title(label="Chunkie sales")

print("Mean of store 1 chunkie sales ",np.mean(store1["chunkie"]))
print("Mean of store 2 chunkie sales ", np.mean(store2["chunkie"]))

#For Honey bite
axes[1,1].hist(store2["honey bite"], label="Store 2")
axes[1,1].hist(store3["honey bite"], alpha=0.6, label="Store 3")
axes[1,1].legend()
axes[1,1].set_title(label="Honey bite sales")

print("Mean of store 2 honey bite sales ",np.mean(store2["honey bite"]))
print("Mean of store 3 honey bite sales ", np.mean(store3["honey bite"]))
```
