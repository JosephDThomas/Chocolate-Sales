# Generating data

```python

import numpy as np
from matplotlib import pyplot as plt
rng = np.random.default_rng()

num_points = 31

x_raw = rng.normal(30, 5, num_points)
y_raw = rng.normal(32, 4, num_points)
z_raw = rng.normal(26, 2, num_points)
sign = rng.choice((-1,1))
noise = rng.poisson(3, num_points)


rng = np.random.default_rng()
arr = rng.choice([-1, 1], size=num_points)

x_noise = np.random.rand(num_points) * 0.15 * noise
y_noise = np.random.rand(num_points) * 0.15 * noise * arr
z_noise = np.random.rand(num_points) * 0.15 * noise * arr

x = np.round(x_raw + x_noise)
y = np.round(y_raw + y_noise) 
z = np.round(z_raw + z_noise)

plt.hist(x, bins=10)
plt.hist(y, bins=10, alpha=0.6)
plt.hist(z, bins=10, alpha=0.6)
```
